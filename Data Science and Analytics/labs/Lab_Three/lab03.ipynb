# 1.a Create DataFrame and Visualize Using Scatter Plot

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create dataset
data = {
    "Area": [1, 1.5, 2, 3, 4, 5],
    "Price": [1.2, 2.3, 2.75, 3.75, 4.25, 4.98]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print(df)

# Scatter Plot
plt.scatter(df["Area"], df["Price"])

# Labels and Title
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price Scatter Plot")

# Show plot
plt.show()
```

# 1.b Apply Linear Regression and determine the equation.

```python
from sklearn.linear_model import LinearRegression

# Create model
model = LinearRegression()

# Independent and dependent variables
X = df[["Area"]]
y = df["Price"]

# Train the model
model.fit(X, y)

# Get slope and intercept
m = model.coef_[0]
c = model.intercept_

print("Slope (m):", m)
print("Intercept (c):", c)

# Equation
print(f"Equation: y = {m:.3f}x + {c:.3f}")
```

# 1.c Calculate MSE and R² Score

```python
from sklearn.metrics import mean_squared_error, r2_score

# Predict values for training data
y_pred = model.predict(X)

# Calculate MSE
mse = mean_squared_error(y, y_pred)

# Calculate R² Score
r2 = r2_score(y, y_pred)

print("Mean Squared Error (MSE):", mse)
print("R² Score:", r2)
```

# 1.d Predict price for new data

```python
# New unseen data
new_data = [[6]]

# Prediction
predicted_price = model.predict(new_data)

print("Predicted price for Area 6:", predicted_price[0])
```

# 1.e Plot the regression line along with the scatter data.

```python
# Plot scatter (actual data)
plt.scatter(X, y, color='blue', label='Actual Data')

# Plot regression line
plt.plot(X, y_pred, color='red', label='Regression Line')

# Labels
plt.xlabel("Area (acres)")
plt.ylabel("Price (lakh)")
plt.title("Regression Line with Scatter Plot")
plt.legend()

# Show plot
plt.show()
```

# 2.a Find the linear regression equation. Also determine its MSE and R² score.

```python
from sklearn.preprocessing import LabelEncoder


# Dataset
data = {
    'Experience': [2, 3, 4, 5, 3],
    'Education': ['High School', 'Bachelor', 'High School', 'Bachelor', 'Master'],
    'Salary': [30000, 48000, 45000, 75000, 70000]
}

df = pd.DataFrame(data)

# Encode Education column
le = LabelEncoder()
df['Education_Code'] = le.fit_transform(df['Education'])

# Features and Target
X = df[['Experience', 'Education_Code']]
y = df['Salary']

# Train model
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

# Calculate MSE
mse = mean_squared_error(y, y_pred)

# Calculate R² Score
r2 = r2_score(y, y_pred)

print("\nMean Squared Error (MSE):", mse)
print("R² Score:", r2)

# Display equation
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

print("\nRegression Equation:")
print(f"Salary = {model.intercept_:.2f} + "
      f"({model.coef_[0]:.2f} × Experience) + "
      f"({model.coef_[1]:.2f} × Education_Code)")
```

# 2.b Create a line of preference (prediction line) to see how well the regression works.



```python
# Predictions
y_pred = model.predict(X)

# Plot Actual vs Predicted Salary
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(y, y_pred)

# Perfect prediction line
plt.plot([y.min(), y.max()],
         [y.min(), y.max()],
         'r--',
         linewidth=2)

plt.xlabel("Actual Salary")
plt.ylabel("Predicted Salary")
plt.title("Actual vs Predicted Salary")
plt.grid(True)

plt.show()
```
