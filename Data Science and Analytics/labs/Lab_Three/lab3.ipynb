Create Dataframe fro the data.


```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Area" : [1,1.5,2,3,4,5],
    "Price" : [1.2,2.3,2.75,3.75,4.25,4.98,]
    
}
df = pd.DataFrame(data)
print(df)

plt.scatter(df["Area"], df["Price"])
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price")
plt.show()
```

b. Apply linear regression Model, Determine its eqution.

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

x = df[["Area"]]
y = df["Price"]

model.fit(x, y)
```
