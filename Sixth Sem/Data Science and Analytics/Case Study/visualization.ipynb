```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("my_data (1).csv")
```

```python
# Plot 1: Age Distribution
plt.figure()
sns.histplot(data=df, x="age", bins=15, kde=True, color="skyblue")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Plot 2: Salary Distribution
plt.figure()
sns.histplot(data=df, x="cgpa", bins=15, kde=True, color="salmon")  # Changed color for variety
plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot 3: Placed Distribution
plt.figure()
sns.countplot(data=df, x="placed", palette="pastel")
plt.title("Placement Distribution")
plt.xlabel("Placed (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()

# Print exact counts and percentage
print(df["placed"].value_counts())
print(df["placed"].value_counts(normalize=True) * 100)
```

```python
# Plot 4: Gender Distribution
plt.figure()
sns.countplot(data=df, x="gender", palette="pastel")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# Plot 5: City Distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="city", palette="pastel", order=df["city"].value_counts().index)
plt.title("City Distribution")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.show()

# Plot 6: Profession Distribution
plt.figure()
sns.countplot(data=df, x="profession", palette="pastel", order=df["profession"].value_counts().index)
plt.title("Profession Distribution")
plt.xlabel("Profession")
plt.ylabel("Count")
plt.show()
```

```python
# Plot 7: Gender vs Placed
plt.figure()
sns.countplot(data=df, x="gender", hue="placed", palette="Set2")
plt.title("Placement by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.legend(title="Placed", labels=["No", "Yes"])
plt.show()

# Plot 8: Profession vs Placed
plt.figure()
sns.countplot(data=df, x="profession", hue="placed", palette="Set2",
              order=df["profession"].value_counts().index)
plt.title("Placement by Profession")
plt.xlabel("Profession")
plt.ylabel("Count")
plt.legend(title="Placed", labels=["No", "Yes"])
plt.show()

# Plot 9: City vs Placed
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="city", hue="placed", palette="Set2",
              order=df["city"].value_counts().index)
plt.title("Placement by City")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.legend(title="Placed", labels=["No", "Yes"])
plt.show()



print("Placement rate by Gender (%):")
print(df.groupby("gender")["placed"].mean() * 100)

print("\nPlacement rate by Profession (%):")
print(df.groupby("profession")["placed"].mean() * 100)

print("\nPlacement rate by City (%):")
print(df.groupby("city")["placed"].mean() * 100)
```

```python
# Plot 10: Age vs Placed
plt.figure()
sns.boxplot(data=df, x="placed", y="age", palette="Set3")
plt.title("Age Distribution by Placement Status")
plt.xlabel("Placed (0 = No, 1 = Yes)")
plt.ylabel("Age")
plt.show()

# Plot 11: CGPA vs Placed
plt.figure()
sns.boxplot(data=df, x="placed", y="cgpa", palette="Set3")
plt.title("CGPA Distribution by Placement Status")
plt.xlabel("Placed (0 = No, 1 = Yes)")
plt.ylabel("CGPA")
plt.show()



print("Age summary by Placement:")
print(df.groupby("placed")["age"].describe())

print("\nCGPA summary by Placement:")
print(df.groupby("placed")["cgpa"].describe())
```

```python
from scipy import stats
```

```python
contingency_table_prof = pd.crosstab(df["profession"], df["placed"])
print("Contingency Table:")
print(contingency_table_prof)

chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table_prof)

print(f"\nChi-square statistic: {chi2_stat:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("\nResult: Reject H0 -> Significant association between Profession and Placement.")
else:
    print("\nResult: Fail to reject H0 -> No significant association between Profession and Placement.")
```
