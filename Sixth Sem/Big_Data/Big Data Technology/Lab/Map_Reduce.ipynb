```python
from collections import defaultdict
```

# === MAP FUNCTION ===

```python
def mapper(text):
    mapped = []
    for line in text.split('\n'):
        words = line.strip().split()
        for word in words:
            mapped.append((word.lower(), 1))
    return mapped
```

# === SHUFFLE & SORT FUNCTION ===

```python
def shuffle_and_sort(mapped):
    shuffled = defaultdict(list)
    for key, value in mapped:
        shuffled[key].append(value)
    return shuffled
```

# === REDUCE FUNCTION ===

```python
def reducer(shuffled):
    reduced = {}
    for key, values in shuffled.items():
        reduced[key] = sum(values)
    return reduced
```

# === SAMPLE TEXT INPUT ===

```python
input_text = """
Nepal is a country
Nepal is situated in Asia
Nepal is a land of natural beauty
Nepal is a land of rich culture
Nepal is a land of rich history
Nepal is a land of rich heritage
"""
```

# === MAP PHASE ===

```python
mapped_output = mapper(input_text)
```

# === SHUFFLE & SORT PHASE ===

```python
shuffled_output = shuffle_and_sort(mapped_output)
```

# === REDUCE PHASE ===

```python
reduced_output = reducer(shuffled_output)
```

# === DISPLAY RESULT ===

```python
print("=== Word Count Result ===")
for word, count in reduced_output.items():
    print(f"{word}: {count}")
```
